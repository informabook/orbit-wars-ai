# Competition and historical results

The project report V5 documents 611 / 4,729 teams (Top 12.92%). This notebook explores local match records. Match attempts are counted within each source. See [research notes](../docs/research-notes.md) and [data dictionary](../DATA_DICTIONARY.md).

```python
from pathlib import Path
import csv
import json
from collections import Counter, defaultdict
import matplotlib.pyplot as plt

root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "results/phase3/EXPERIMENT_MANIFEST.json").is_file())
results = root / "results"

def load_csv(relative):
    with (results / relative).open(encoding="utf-8", newline="") as handle:
        return list(csv.DictReader(handle))
```

```python
local = load_csv("phase3/orbitlab_individual_matches.csv")
evolution = load_csv("evolution_matches.csv")
mode_status = Counter((r["mode"], r["status"]) for r in local)
modes = sorted({r["mode"] for r in local})
fig, ax = plt.subplots()
completed = [mode_status[m, "OK"] for m in modes]
errors = [mode_status[m, "ERROR"] for m in modes]
ax.bar(modes, completed, label="Recorded OK")
ax.bar(modes, errors, bottom=completed, label="Recorded ERROR")
ax.set(ylabel="Match records", title="OrbitLab outputs by game mode")
ax.legend()
plt.show()
print("OrbitLab:", len(local))
print("Evolution journal:", Counter(r["status"] for r in evolution))
```

The OrbitLab records carry archived OK/ERROR flags. They lack per-agent terminal statuses. The logs use separate ID schemes and the full historical campaign coverage remains unknown.
